# Turkish AI evaluation tables

This notebook creates the three tables from the JSON files in `data/university_entrance/`.
Run all cells to load the current data and display every row. Rates are displayed
as percentages with two decimal places; the underlying DataFrames retain the
original numeric values, column names, and row order.


```python
from pathlib import Path

import pandas as pd
from IPython.display import display

from tr_ai_evals.explorer.results import RATE_COLUMNS, load_table

# Locate the project when the kernel starts in the root or a subdirectory.
DATA_FILES = (
    "overall_comparison.json",
    "course_accuracy_comparison.json",
    "course_detailed_metrics.json",
)
PROJECT_ROOT = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if all((path / "data" / "university_entrance" / name).is_file() for name in DATA_FILES)),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError("Start the notebook inside the project containing data/university_entrance/.")
DATA_DIR = PROJECT_ROOT / "data" / "university_entrance"


def style_table(table, caption, percentage_columns):
    formats = {column: "{:.2%}" for column in percentage_columns}
    formats.update({column: "{:,}" for column in
                    ("Questions", "Correct", "Incorrect", "Invalid / missing")
                    if column in table.columns})
    return (
        table.style
        .format(formats)
        .hide(axis="index")
        .set_caption(caption)
        .set_table_styles([
            {"selector": "caption", "props": [("text-align", "left"),
                                               ("font-weight", "bold")]},
            {"selector": "th", "props": [("text-align", "left")]},
            {"selector": "td", "props": [("padding", "6px 10px")]},
        ])
    )
```

## 1. Overall comparison

One row per experiment and model, loaded from `overall_comparison.json`.
The source's **Incorrect** count includes invalid or missing predictions;
**Invalid / missing** is a subset, not an additional category to sum.


```python
overall_comparison = load_table(DATA_DIR / "overall_comparison.json")
display(style_table(overall_comparison, "Overall comparison", RATE_COLUMNS))
```

## 2. Course accuracy comparison

One row per course, with question counts and accuracy for each experiment,
loaded from `course_accuracy_comparison.json`.


```python
course_accuracy_comparison = load_table(DATA_DIR / "course_accuracy_comparison.json")
experiment_columns = [column for column in course_accuracy_comparison.columns
                      if column.startswith("Exp ")]
display(style_table(course_accuracy_comparison, "Course accuracy comparison", experiment_columns))
```

## 3. Detailed course metrics

One row per course and experiment, loaded from `course_detailed_metrics.json`.
All rows are displayed. As in the overall table, **Incorrect** includes
invalid or missing predictions.


```python
course_detailed_metrics = load_table(DATA_DIR / "course_detailed_metrics.json")
display(style_table(course_detailed_metrics, "Detailed course metrics", RATE_COLUMNS))
```
